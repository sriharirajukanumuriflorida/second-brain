## 3. Add a guardrail

```python
def guarded(fn, budget=3):
    out=[]
    for i in range(budget):
        out.append(fn(i))
    return out
print(guarded(lambda i: f"step-{i}", 3))
```

## 4. Deterministic scoring

```python
scores=np.array([3.0,2.0,1.0]); probs=np.exp(scores)/np.exp(scores).sum()
print(np.round(probs,3), "choice", int(probs.argmax()))
```
