## 1. Minimal simulation

```python
class Budget:
    def __init__(self): self.steps=0
    def charge(self): self.steps+=1; assert self.steps<=3
b=Budget()
for _ in range(3): b.charge(); log("step", n=b.steps)
print(b.steps)
```

## 3. Add a guardrail

```python
def guarded(fn, budget=3):
    out=[]
    for i in range(budget):
        out.append(fn(i))
    return out
print(guarded(lambda i: f"step-{i}", 3))
```

## 4. Deterministic scoring

```python
scores=np.array([3.0,2.0,1.0]); probs=np.exp(scores)/np.exp(scores).sum()
print(np.round(probs,3), "choice", int(probs.argmax()))
```
