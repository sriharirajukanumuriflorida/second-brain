# Preference Optimization — Hands-On

## 0. Setup

```python
%pip install -q numpy
import numpy as np
def logsigmoid(x): return -np.logaddexp(0,-x)
```

## 1. DPO loss

```python
pi_c=np.array([-7.0,-4.0,-9.0]); pi_r=np.array([-8.0,-5.5,-8.8]); ref_c=np.array([-7.5,-4.8,-9.1]); ref_r=np.array([-8.1,-5.0,-8.9]); beta=.5
margin=(pi_c-pi_r)-(ref_c-ref_r); loss=-logsigmoid(beta*margin)
print(np.round(margin,3)); print(np.round(loss,3)); print('mean', round(float(loss.mean()),3))
```

## 2. Preference pressure

```python
for b in [.1,.5,1,2]: print(b, round(float((-logsigmoid(b*margin)).mean()),3))
```

## 3. Reward model toy accuracy

```python
features=np.array([[1,0],[1,1],[0,1],[2,1]],float); w=np.array([.8,.3]); chosen=features@w; rejected=np.array([.1,.5,.4,1.0]); print('pair accuracy', float((chosen>rejected).mean()))
```

## 4. Pair quality checks

```python
pairs=[('short','short'),('policy answer','made up guarantee'),('accurate','very very very long but wrong')]
for c,r in pairs: print('duplicate', c==r, 'length_gap', abs(len(c)-len(r)))
```

## 5. Reward hacking proxy

```python
answers=['concise correct','long flattering maybe wrong','refuses safely']; reward=np.array([.8,.9,.6]); task=np.array([1.0,.3,.7]); print(answers[int(reward.argmax())], 'reward winner'); print(answers[int(task.argmax())], 'task winner')
```

## 6. ORPO intuition

```python
sft_loss=np.array([.4,.5,.7]); odds_penalty=np.log1p(np.exp(-(pi_c-pi_r))); print(np.round(sft_loss+.2*odds_penalty,3))
```
