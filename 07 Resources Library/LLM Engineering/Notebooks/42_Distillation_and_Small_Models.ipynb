# Distillation & Small Models — Hands-On

## 0. Setup

```python
%pip install -q numpy
import numpy as np
def softmax(z,T=1):
    z=np.array(z,dtype=float)/T; e=np.exp(z-z.max(axis=-1,keepdims=True)); return e/e.sum(axis=-1,keepdims=True)
```

## 1. Teacher and student distributions

```python
teacher_logits=np.array([[4,2,0],[1,3,0]],float); student_logits=np.array([[2,1,1],[1,1.5,.5]],float)
pt=softmax(teacher_logits,2); ps=softmax(student_logits,2); print(np.round(pt,3)); print(np.round(ps,3))
```

## 2. KL loss

```python
kl=np.sum(pt*(np.log(pt)-np.log(ps)),axis=1); print(np.round(kl,4), 'mean', round(float(kl.mean()),4))
```

## 3. Temperature changes dark knowledge

```python
for T in [1,2,4]: print(T, np.round(softmax(teacher_logits[0],T),3))
```

## 4. Sequence-level distillation data

```python
prompts=['refund?','summarize']; teacher_answers=['Ask for order id and cite policy.','Return a concise factual summary.']; print(list(zip(prompts,teacher_answers)))
```

## 5. Cost latency tradeoff

```python
acc=np.array([.82,.86,.9]); latency=np.array([40,120,800]); cost=np.array([.02,.08,.8]); utility=acc-.0002*latency-.2*cost; print(np.round(utility,3), 'best', int(utility.argmax()))
```

## 6. Escalation router

```python
def route(conf,risk): return 'teacher' if risk=='high' or conf<.72 else 'student'
for c,r in [(0.9,'low'),(.6,'low'),(.8,'high')]: print(c,r,route(c,r))
```
