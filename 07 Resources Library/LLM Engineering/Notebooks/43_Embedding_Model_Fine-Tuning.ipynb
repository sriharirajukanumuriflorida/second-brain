# Embedding-Model Fine-Tuning — Hands-On

## 0. Setup

```python
%pip install -q numpy
import numpy as np
def norm(x): return x/np.linalg.norm(x,axis=1,keepdims=True)
A=norm(np.array([[1,0,0],[0,1,0],[1,1,0]],float)); P=norm(np.array([[.9,.1,0],[.1,.9,0],[.7,.7,.1]],float))
```

## 1. Similarity matrix

```python
sim=A@P.T; print(np.round(sim,3))
```

## 2. MNRL loss

```python
temp=.1; z=sim/temp; exp=np.exp(z-z.max(1,keepdims=True)); prob=exp/exp.sum(1,keepdims=True); loss=-np.log(np.diag(prob)).mean(); print(np.round(prob,3)); print(round(float(loss),3))
```

## 3. Hard negatives

```python
for i,row in enumerate(sim):
    r=row.copy(); r[i]=-1
    print(i, 'hard negative doc', int(r.argmax()), 'score', round(float(r.max()),3))
```

## 4. Triplet margin

```python
neg=P[[1,2,0]]; pos=np.sum(A*P,1); negs=np.sum(A*neg,1); margin=.2; trip=np.maximum(0, margin - pos + negs); print(np.round(trip,3))
```

## 5. Retrieval metrics

```python
rankings=np.array([[0,1,2],[2,1,0],[1,2,0]]); gold=[0,1,2]
recall1=np.mean([gold[i] in rankings[i,:1] for i in range(3)]); mrr=np.mean([1/(list(rankings[i]).index(gold[i])+1) for i in range(3)])
print('recall@1', recall1, 'mrr', round(float(mrr),3))
```

## 6. Re-index warning

```python
old=np.array([1,0]); new=np.array([0,1]); print('same stored vector under new model?', np.allclose(old,new))
```
